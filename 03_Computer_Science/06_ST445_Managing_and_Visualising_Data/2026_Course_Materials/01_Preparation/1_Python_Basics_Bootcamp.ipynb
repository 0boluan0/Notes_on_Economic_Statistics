## 1. Variables and Data Types

### 📖 Explanation
A **variable** is a name that refers to a value stored in memory. Python is **dynamically typed**, which means you don't declare a type. Python works out the type from the value you assign.

### 🧩 Syntax
```python
variable_name = value
```
| Part | Meaning |
|---|---|
| `variable_name` | The label. It must start with a letter or `_`, may contain letters, digits, and `_`, and is **case-sensitive** (`Age` ≠ `age`). |
| `=` | The **assignment** operator. It is not equality. Use `==` to compare values. |
| `value` | Any Python object, such as a number, text, a list, and so on. |

**Naming convention:** use `snake_case` (for example `total_sales`, `user_id`).

### Core built-in data types
| Type | Name | Example | Used for |
|---|---|---|---|
| `int` | Integer | `42`, `-7` | Counts, IDs |
| `float` | Floating-point | `3.14`, `2.5e3` | Measurements, prices |
| `str` | String | `"hello"`, `'data'` | Text, column names, HTML |
| `bool` | Boolean | `True`, `False` | Conditions and filters |
| `NoneType` | None | `None` | "No value" or missing data |
| `list`, `tuple`, `set`, `dict` | Collections | covered later | Groups of values |

**Useful built-in functions**
- `type(x)` returns the type of `x`.
- `isinstance(x, type)` returns `True` if `x` is of that type.
- `print(*values, sep=' ', end='\n')` displays values. `sep` is placed between the values and `end` is added after the last one.

````python
# Creating variables of different types
student_name = "Ada"        # str
age = 28                    # int
gpa = 3.85                  # float
is_enrolled = True          # bool
middle_name = None          # NoneType

print(student_name, age, gpa, is_enrolled, middle_name)
print(type(student_name), type(age), type(gpa), type(is_enrolled), type(middle_name))
````

````python
# Multiple assignment and swapping
x, y, z = 1, 2, 3
print(x, y, z)

x, y = y, x          # swap without a temporary variable
print("After swap:", x, y)

# The print() parameters sep and end
print("2026", "10", "05", sep="-")
print("No newline here...", end=" ")
print("continues on the same line.")
````

````python
# Arithmetic operators
a, b = 17, 5
print("Addition:       ", a + b)
print("Subtraction:    ", a - b)
print("Multiplication: ", a * b)
print("Division:       ", a / b)    # always returns a float
print("Floor division: ", a // b)   # drops the decimal part
print("Modulus:        ", a % b)    # remainder
print("Exponent:       ", a ** b)

# Augmented assignment
count = 10
count += 5     # same as count = count + 5
print("count:", count)
print(isinstance(count, int))
````

## 2. Casting (Type Conversion)

### 📖 Explanation
**Casting** converts a value from one type to another. This is essential in data work. Data from files, APIs, and web pages often arrives **as text** even when it represents numbers, for example `"42"` or `"19.99"`.

### 🧩 Syntax
| Function | Converts to | Parameters and notes |
|---|---|---|
| `int(x, base=10)` | Integer | `x` is a number or numeric string. `base` is the number base for strings (for example `int("ff", 16)`). Floats are **truncated**, not rounded. |
| `float(x)` | Float | `x` is a number or string such as `"3.5"`, `"1e3"`, `"nan"`. |
| `str(x)` | String | Works on almost any object. |
| `bool(x)` | Boolean | `0`, `0.0`, `""`, `[]`, `{}`, and `None` become `False`. Everything else becomes `True`. |
| `round(number, ndigits=None)` | Rounded number | `ndigits` is the number of decimal places to keep. |

⚠️ Casting fails with a `ValueError` when the conversion makes no sense, for example `int("hello")`. You'll learn to handle this in **Section 11**.

````python
price_text = "19.99"       # e.g., scraped from a web page
quantity_text = "3"

# "19.99" * "3" would fail, so cast first
total = float(price_text) * int(quantity_text)
print("Total:", total)
print("Rounded:", round(total, 1))

print(int(7.9))            # 7  (truncated, not rounded)
print(int("101", 2))       # 5  (binary string to int)
print(str(2026) + " is the year")

print(bool(0), bool(""), bool("False"), bool([]), bool([0]))
````

````python
# Implicit conversion: Python automatically promotes int to float in mixed math
result = 5 + 2.0
print(result, type(result))

# True counts as 1 and False as 0, which is handy for counting matches
print(True + True + False)
````

## 3. Strings

### 📖 Explanation
A **string** (`str`) is an ordered, **immutable** sequence of characters. Immutable means it can't be changed in place, so string methods always return a **new** string. Strings are everywhere in data work: column names, CSV rows, JSON fields, URLs, and HTML.

### 🧩 Syntax
```python
s = "double quotes"    # or 'single quotes'
s = """triple quotes
span multiple lines"""
```

**Indexing and slicing**: `s[start:stop:step]`
| Parameter | Meaning | Default |
|---|---|---|
| `start` | First index included (0-based) | `0` |
| `stop` | Index where the slice stops (**not included**) | end of string |
| `step` | Gap between indices (negative values go backwards) | `1` |

Negative indices count from the end: `s[-1]` is the last character.

**Common methods** (`s` is a string)
| Method | What it does |
|---|---|
| `s.lower()` / `s.upper()` / `s.title()` | Change the case |
| `s.strip(chars=None)` | Remove whitespace (or the given `chars`) from both ends. Variants: `lstrip`, `rstrip` |
| `s.split(sep=None, maxsplit=-1)` | Break into a **list** at `sep` (default: any whitespace). `maxsplit` limits the number of splits |
| `sep.join(iterable)` | Glue a list of strings together with `sep` between them |
| `s.replace(old, new, count=-1)` | Replace `old` with `new`. `count` limits the replacements |
| `s.find(sub)` | Index of the first `sub`, or `-1` if not found |
| `s.startswith(prefix)` / `s.endswith(suffix)` | Return a bool |
| `s.isdigit()` / `s.isalpha()` | Check the characters it contains |
| `len(s)` | Number of characters (a built-in function, not a method) |
| `sub in s` | `True` if `sub` appears in `s` |

**f-strings** (formatted strings) insert values directly into text: `f"{value:format_spec}"`
- `{x:.2f}` gives 2 decimal places, `{x:,}` adds thousands separators, `{x:>10}` right-aligns in 10 characters, and `{x:.1%}` displays a percentage.

````python
title = "  Introduction to Web Scraping  "

print(repr(title.strip()))      # repr() shows the quotes, which makes spaces visible
print(title.strip().upper())
print(title.strip().lower().replace(" ", "_"))   # a typical "clean column name" pattern
print(len(title), len(title.strip()))
````

````python
word = "pandas"
print(word[0])      # 'p'
print(word[-1])     # 's'
print(word[1:4])    # 'and' -> indices 1, 2, 3
print(word[:3])     # 'pan'
print(word[::2])    # 'pna' -> every second character
print(word[::-1])   # 'sadnap' -> reversed
````

````python
# split and join: the bread and butter of parsing text data
csv_line = "Ada,28,London,3.85"
fields = csv_line.split(",")
print(fields)

url_parts = ["https:", "", "api.example.com", "v1", "users"]
print("/".join(url_parts))

# f-strings with format specifiers
name, score, revenue, rate = "Grace", 91.4567, 1234567.891, 0.0753
print(f"{name} scored {score:.2f}")
print(f"Revenue: ${revenue:,.2f}")
print(f"Growth rate: {rate:.1%}")
print(f"|{name:>10}|{name:<10}|")

# Membership and searching
html = "<div class='price'>$19.99</div>"
print("price" in html, html.find("$"), html.startswith("<div"))
````

## 4. Lists

### 📖 Explanation
A **list** is an ordered, **mutable** (changeable) collection that can hold any mix of types and allows duplicates. Lists are Python's general-purpose container. A pandas column, a JSON array, and the rows scraped from a table all map naturally onto lists.

### 🧩 Syntax
```python
my_list = [item1, item2, item3]
empty = []            # or list()
```
Indexing and slicing work exactly like strings: `my_list[start:stop:step]`.

**Common methods** (they modify the list **in place** and return `None`, unless stated otherwise)
| Method | What it does |
|---|---|
| `lst.append(x)` | Add `x` to the end |
| `lst.extend(iterable)` | Add every item from `iterable` |
| `lst.insert(i, x)` | Insert `x` at index `i` |
| `lst.remove(x)` | Remove the first occurrence of `x` (raises `ValueError` if it isn't there) |
| `lst.pop(i=-1)` | Remove **and return** the item at index `i` (default: the last item) |
| `lst.sort(key=None, reverse=False)` | Sort in place. `key` is a function applied to each item before comparing, and `reverse=True` sorts in descending order |
| `sorted(iterable, key=None, reverse=False)` | Built-in function that returns a **new** sorted list |
| `lst.index(x)` / `lst.count(x)` | Position of `x` / number of times `x` appears |
| `len(lst)`, `sum(lst)`, `min(lst)`, `max(lst)` | Built-in summaries |

⚠️ `b = a` does **not** copy a list. Both names point to the same object. Use `b = a.copy()` or `b = a[:]` to make a copy.

````python
scores = [88, 92, 79, 92, 65]
print(scores[0], scores[-1], scores[1:3])

scores.append(100)
scores.insert(0, 50)
print("After append/insert:", scores)

removed = scores.pop()        # removes and returns 100
scores.remove(92)             # removes only the FIRST 92
print("Popped:", removed, "| Now:", scores)

scores[0] = 55                # lists are mutable
print("Modified:", scores)
````

````python
nums = [4, 1, 3, 2]
print(sorted(nums), nums)          # sorted() returns a new list; the original is unchanged
nums.sort(reverse=True)
print(nums)                        # sort() changed the list itself

words = ["banana", "Kiwi", "apple"]
print(sorted(words, key=str.lower))   # key: sort ignoring case
print(sorted(words, key=len))         # key: sort by length

print(len(nums), sum(nums), min(nums), max(nums), sum(nums) / len(nums))
````

````python
# Nested lists act like a table (rows x columns), much like a mini DataFrame
table = [
    ["name", "age"],
    ["Ada", 28],
    ["Linus", 35],
]
print(table[1])       # second row
print(table[2][0])    # row 2, column 0 -> 'Linus'

# The copying trap
a = [1, 2, 3]
b = a           # same object!
c = a.copy()    # independent copy
a.append(4)
print("a:", a, "| b:", b, "| c:", c)
````

## 5. Tuples

### 📖 Explanation
A **tuple** is an ordered, **immutable** collection. Once created, it can't be changed. Use tuples for fixed groups of values that belong together, such as coordinates, `(row, column)` pairs, or records. You will see them often: `df.shape` in pandas returns a tuple like `(1000, 5)`, and functions can return several values as a tuple.

### 🧩 Syntax
```python
point = (3, 4)
single = (5,)        # a one-item tuple NEEDS the trailing comma
also_tuple = 3, 4    # parentheses are optional
```
| Operation | Meaning |
|---|---|
| `t[i]`, `t[a:b]` | Indexing and slicing (the same as lists) |
| `t.count(x)` | Number of times `x` appears |
| `t.index(x)` | Position of the first `x` |
| `a, b = t` | **Unpacking**: assign each item to its own variable |
| `first, *rest = t` | Starred unpacking: `rest` collects the remaining items as a list |

**Why choose tuples over lists?** They protect data from accidental changes, they are slightly faster, and they can be used as dictionary keys or set members (lists can't).

````python
location = (51.5074, -0.1278)      # (latitude, longitude)
print(location[0], type(location))

lat, lon = location                 # unpacking
print(f"Lat: {lat}, Lon: {lon}")

single = (5,)
not_tuple = (5)
print(type(single), type(not_tuple))

first, *rest = (10, 20, 30, 40)
print(first, rest)

shape = (1000, 5)                    # like pandas df.shape
rows, cols = shape
print(f"{rows} rows x {cols} columns")
````

````python
# Tuples are immutable, so this raises an error (we'll learn to catch it in Section 11)
try:
    location[0] = 0
except TypeError as e:
    print("TypeError:", e)
````

## 6. Sets

### 📖 Explanation
A **set** is an **unordered** collection of **unique** items. Duplicates are removed automatically. Sets are ideal for removing duplicates, testing membership quickly, and comparing groups (for example, "which user IDs appear in both datasets?").

### 🧩 Syntax
```python
s = {1, 2, 3}
empty = set()        # NOT {}: that creates an empty dictionary
```
| Method / Operator | What it does |
|---|---|
| `s.add(x)` | Add one item |
| `s.update(iterable)` | Add several items |
| `s.remove(x)` | Remove `x` (raises `KeyError` if missing) |
| `s.discard(x)` | Remove `x` if present (no error) |
| `a \| b` or `a.union(b)` | Items in **either** set |
| `a & b` or `a.intersection(b)` | Items in **both** sets |
| `a - b` or `a.difference(b)` | Items in `a` but **not** in `b` |
| `a ^ b` or `a.symmetric_difference(b)` | Items in exactly **one** of the sets |
| `x in s` | Very fast membership test |

⚠️ Sets have no index, so `s[0]` is an error. Only immutable items (numbers, strings, tuples) can go in a set.

````python
tags = ["python", "data", "python", "api", "data", "scraping"]
unique_tags = set(tags)
print(unique_tags, len(unique_tags))

unique_tags.add("numpy")
unique_tags.discard("not-there")   # no error
print(unique_tags)
````

````python
dataset_a_ids = {101, 102, 103, 104}
dataset_b_ids = {103, 104, 105}

print("In both:        ", dataset_a_ids & dataset_b_ids)
print("In either:      ", dataset_a_ids | dataset_b_ids)
print("Only in A:      ", dataset_a_ids - dataset_b_ids)
print("In exactly one: ", dataset_a_ids ^ dataset_b_ids)
print(105 in dataset_b_ids)
````

## 7. Dictionaries

### 📖 Explanation
A **dictionary** (`dict`) stores **key → value** pairs. It is *the* most important structure for API work, because **JSON data maps directly to Python dictionaries and lists**. A pandas DataFrame can also be built straight from a dict of lists.

Keys must be unique and immutable (strings, numbers, tuples). Values can be anything, including other dictionaries and lists. Since Python 3.7, dictionaries keep their insertion order.

### 🧩 Syntax
```python
d = {"key1": value1, "key2": value2}
empty = {}           # or dict()
```
| Method / Operation | What it does |
|---|---|
| `d[key]` | Get a value (raises `KeyError` if the key is missing) |
| `d.get(key, default=None)` | Get a value, or return `default` if the key is missing (**safe for API data**) |
| `d[key] = value` | Add or update an entry |
| `d.keys()` / `d.values()` / `d.items()` | Views of the keys, the values, or the `(key, value)` pairs |
| `d.update(other)` | Merge in another dictionary |
| `d.pop(key, default)` | Remove a key and return its value |
| `key in d` | Check whether a key exists |

````python
user = {
    "id": 42,
    "name": "Ada Lovelace",
    "skills": ["math", "programming"],
    "address": {"city": "London", "country": "UK"},
}

print(user["name"])
print(user["address"]["city"])          # nested access, very common with JSON
print(user["skills"][1])
print(user.get("email"))                # None instead of a crash
print(user.get("email", "not provided"))

user["email"] = "ada@example.com"       # add a key
user["id"] = 43                          # update a key
print(list(user.keys()))
print("skills" in user)
````

````python
# A typical API response shape: a dictionary containing a list of dictionaries
api_response = {
    "status": "ok",
    "count": 3,
    "results": [
        {"city": "Paris", "temp_c": 18.2},
        {"city": "Tokyo", "temp_c": 22.5},
        {"city": "Lima", "temp_c": 16.9},
    ],
}

for item in api_response["results"]:
    print(f"{item['city']:<6} {item['temp_c']}°C")

# A dict of lists is exactly what pandas.DataFrame() accepts
table = {"city": ["Paris", "Tokyo"], "temp_c": [18.2, 22.5]}
print(table)
````

## 8. Control Structures: `if`, `elif`, `else`

### 📖 Explanation
**Conditional statements** let your code make decisions: run one block if a condition is `True` and a different block otherwise. In data work you'll use them to validate inputs, check API status codes, and categorize values.

### 🧩 Syntax
```python
if condition1:
    # runs if condition1 is True
elif condition2:
    # runs if condition1 is False and condition2 is True
else:
    # runs if none of the above were True
```
- The colon `:` ends the header line.
- **Indentation (4 spaces)** defines the block. Python uses it instead of `{ }`.
- `elif` and `else` are optional. You can have any number of `elif` branches.

**Comparison operators:** `==` (equal), `!=` (not equal), `<`, `>`, `<=`, `>=`
**Logical operators:** `and` (both true), `or` (at least one true), `not` (reverses the result)
**Membership and identity:** `in`, `not in`, and `is` / `is not` (use `is` only for `None`, as in `x is None`)

**Conditional (ternary) expression**, a one-line `if`:
```python
result = value_if_true if condition else value_if_false
```

**Truthiness:** empty values (`""`, `[]`, `{}`, `0`, `None`) count as `False`, so `if my_list:` means "if the list isn't empty".

````python
status_code = 404   # try changing this to 200, 301, 500

if status_code == 200:
    print("Success: data received")
elif 300 <= status_code < 400:      # chained comparison
    print("Redirect")
elif status_code == 404:
    print("Not found: check the URL")
else:
    print("Some other problem:", status_code)
````

````python
temp = 31
humidity = 75

if temp > 30 and humidity > 70:
    print("Hot and humid")

if not (temp < 0):
    print("Not freezing")

label = "warm" if temp >= 20 else "cool"     # ternary expression
print(label)

results = []
if results:
    print("We have results")
else:
    print("No results (the empty list is falsy)")

value = None
if value is None:
    print("Missing value detected")
````

## 9. Loops

### 📖 Explanation
**Loops** repeat a block of code. A `for` loop goes through each item of a sequence (an *iterable*). A `while` loop repeats as long as a condition stays `True`. You'll loop over rows, files, API pages, and scraped HTML elements.

> 💡 With pandas and NumPy, you'll often replace explicit loops with **vectorized** operations, which are much faster. Understanding loops is still essential for knowing what those operations do.

### 🧩 Syntax
```python
for item in iterable:
    # body runs once per item

while condition:
    # body repeats while condition is True
```

**Helpful built-ins**
| Function | Parameters | Produces |
|---|---|---|
| `range(start, stop, step)` | `start` (default 0), `stop` (**excluded**), `step` (default 1) | A sequence of integers |
| `enumerate(iterable, start=0)` | `start` is the first counter value | `(index, item)` pairs |
| `zip(*iterables)` | Two or more iterables | Tuples pairing up the items. Stops at the shortest iterable |
| `dict.items()` | — | `(key, value)` pairs |

**Loop control**
- `break` exits the loop immediately.
- `continue` skips the rest of the current pass and moves on to the next one.
- `else` on a loop runs only if the loop finished **without** `break`.

**List comprehension**, a compact way to build a list:
```python
new_list = [expression for item in iterable if condition]
```

````python
for i in range(5):            # 0, 1, 2, 3, 4
    print(i, end=" ")
print()

for i in range(2, 11, 2):     # start=2, stop=11 (excluded), step=2
    print(i, end=" ")
print()

for ch in "data":
    print(ch.upper(), end=" ")
print()
````

## 10. Functions

### 📖 Explanation
A **function** is a named, reusable block of code. Functions keep code organized, easy to test, and free of repetition. In data work you'll write functions to clean values, fetch a URL, or parse a page. You'll also pass functions to pandas methods such as `df.apply(my_function)`.

### 🧩 Syntax
```python
def function_name(param1, param2=default_value, *args, **kwargs):
    """Docstring: describes what the function does."""
    # body
    return result
```
| Part | Meaning |
|---|---|
| `def` | Keyword that starts the function definition |
| `param1` | **Required (positional) parameter**: the caller must supply it |
| `param2=default_value` | **Optional parameter** with a default. Defaults must come after the required parameters |
| `*args` | Collects any extra positional arguments into a **tuple** |
| `**kwargs` | Collects any extra keyword arguments into a **dict** |
| `"""Docstring"""` | Documentation. View it with `help(function_name)` |
| `return` | Sends a value back to the caller. Without it, the function returns `None` |

**Parameters vs. arguments:** *parameters* are the names in the definition, and *arguments* are the actual values passed when you call the function.
**Calling styles:** positional `f(1, 2)`, keyword `f(param2=2, param1=1)`, or a mix (positional arguments first).

**Scope:** variables created inside a function are **local**, meaning they don't exist outside it.

**Lambda (anonymous) function**, a one-line function:
```python
lambda parameters: expression
```
**Type hints** (optional) document the expected types: `def f(x: int) -> float:`

````python
def greet(name, greeting="Hello"):
    """Return a greeting for the given name."""
    return f"{greeting}, {name}!"

print(greet("Ada"))                       # uses the default greeting
print(greet("Ada", "Welcome"))            # positional
print(greet(greeting="Hi", name="Linus")) # keyword arguments, in any order
help(greet)
````

````python
def celsius_to_fahrenheit(c: float) -> float:
    """Convert Celsius to Fahrenheit."""
    return c * 9 / 5 + 32

def summarize(values):
    """Return the min, max, and mean of a list as a tuple."""
    return min(values), max(values), sum(values) / len(values)

print(celsius_to_fahrenheit(25))

lo, hi, avg = summarize([3, 9, 4, 8])     # returning several values = returning a tuple
print(lo, hi, round(avg, 2))
````

## 11. Error Handling

### 📖 Explanation
Errors at runtime are called **exceptions**. Without handling, an exception stops your program. Real-world data is messy: an API might time out, a web page might be missing an element, or a value might be `"N/A"` instead of a number. **Error handling** lets your code respond gracefully instead of crashing.

### 🧩 Syntax
```python
try:
    # code that might fail
except SpecificError as e:
    # runs if SpecificError happens; e holds the error details
except (ErrorA, ErrorB):
    # handle several error types the same way
else:
    # runs only if NO exception occurred
finally:
    # ALWAYS runs (cleanup: closing files, connections)
```
**Raising your own error:** `raise ValueError("helpful message")`

**Common exceptions**
| Exception | Typical cause |
|---|---|
| `ValueError` | Right type, wrong value: `int("abc")` |
| `TypeError` | Wrong type: `"a" + 1` |
| `KeyError` | Missing dictionary key |
| `IndexError` | List index out of range |
| `ZeroDivisionError` | Dividing by zero |
| `FileNotFoundError` | The file path doesn't exist |
| `AttributeError` | The object has no such attribute or method (common in scraping when an element is `None`) |
| `ModuleNotFoundError` | The package isn't installed |

✅ **Best practice:** catch *specific* exceptions. A bare `except:` hides bugs.

````python
def safe_divide(a, b):
    try:
        result = a / b
    except ZeroDivisionError:
        print("Cannot divide by zero; returning None")
        return None
    except TypeError as e:
        print("Type problem:", e)
        return None
    else:
        print("Division succeeded")
        return result
    finally:
        print("-- safe_divide finished --")

print(safe_divide(10, 2))
print(safe_divide(10, 0))
print(safe_divide(10, "2"))
````

````python
# A practical pattern: convert messy values, using None for anything unparseable
raw_values = ["12", "7.5", "N/A", "", "42", None]
parsed = []
for v in raw_values:
    try:
        parsed.append(float(v))
    except (ValueError, TypeError):
        parsed.append(None)
print(parsed)
````

````python
# Raising errors to enforce rules
def set_age(age):
    if not isinstance(age, int):
        raise TypeError("age must be an integer")
    if age < 0:
        raise ValueError(f"age cannot be negative (got {age})")
    return age

for test in [25, -3, "thirty"]:
    try:
        print("Valid age:", set_age(test))
    except (TypeError, ValueError) as e:
        print(f"{type(e).__name__}: {e}")
````

## 12. File Loading (Reading and Writing Files)

### 📖 Explanation
Data usually lives in files: **.txt**, **.csv**, and **.json**. Python reads them with the built-in `open()` function, and the `csv` and `json` modules help with structured formats. Later, pandas will do much of this in one line (`pd.read_csv`), but knowing the basics helps when files are unusual or messy.

### 🧩 Syntax
```python
with open(file, mode="r", encoding="utf-8") as f:
    content = f.read()
```
| Parameter | Meaning |
|---|---|
| `file` | Path to the file, for example `"data.csv"` or `"/content/drive/MyDrive/data.csv"` |
| `mode` | `"r"` read (default), `"w"` write (**overwrites**), `"a"` append, `"x"` create (fails if the file exists), `"b"` binary (for example `"rb"`) |
| `encoding` | The text encoding. `"utf-8"` is the safe standard |
| `newline` | Use `newline=""` when writing CSV files to avoid blank lines |

**Why `with`?** It is a *context manager* that closes the file automatically, even if an error occurs.

**Reading methods:** `f.read()` returns the whole file as one string, `f.readline()` returns one line, `f.readlines()` returns a list of lines, and `for line in f:` reads line by line (memory-friendly).
**Writing methods:** `f.write(string)` and `f.writelines(list_of_strings)`.

**Helper modules**
| Function | Purpose |
|---|---|
| `csv.reader(f, delimiter=",")` | Yields each row as a list |
| `csv.DictReader(f)` | Yields each row as a dict, using the header row as keys |
| `csv.writer(f).writerow(list)` | Writes one row |
| `json.load(f)` / `json.dump(obj, f, indent=2)` | Read JSON from a file / write JSON to a file |
| `json.loads(string)` / `json.dumps(obj)` | Convert between a JSON **string** and Python objects (used with APIs) |

## 13. Arrays: the `array` Module and an Introduction to NumPy

### 📖 Explanation
A Python **list** can hold mixed types, which makes it flexible but slow for heavy math. An **array** stores items of **one type** packed tightly in memory.

- Python's built-in **`array` module** provides basic typed arrays.
- **NumPy's `ndarray`** is the array type used in data science. It supports multiple dimensions and **vectorized math**: operations apply to every element at once, with no loop needed. pandas is built on top of NumPy.

### 🧩 Syntax: the `array` module
```python
from array import array
arr = array(typecode, initializer)
```
| Parameter | Meaning |
|---|---|
| `typecode` | The element type: `'i'` signed int, `'f'` float, `'d'` double (high-precision float), `'u'` unicode character |
| `initializer` | An optional list or iterable of starting values |

### 🧩 Syntax: NumPy essentials
```python
import numpy as np
a = np.array(object, dtype=None)
```
| Function / Attribute | Parameters and meaning |
|---|---|
| `np.array(object, dtype=None)` | Create an array from a list (or a nested list for 2-D). `dtype` forces a type, for example `float` |
| `np.arange(start, stop, step)` | Like `range`, but returns an array |
| `np.linspace(start, stop, num=50)` | `num` evenly spaced values, **including** `stop` |
| `np.zeros(shape)` / `np.ones(shape)` | Arrays filled with 0 or 1. `shape` is an int or a tuple such as `(2, 3)` |
| `a.shape`, `a.ndim`, `a.dtype`, `a.size` | Dimensions (a tuple), number of axes, element type, total number of elements |
| `a.reshape(new_shape)` | The same data with a new shape |
| `a.sum(axis=None)`, `a.mean()`, `a.max()`, `a.std()` | Aggregations. `axis=0` works down the columns and `axis=1` across the rows |
| `a[a > 5]` | **Boolean masking**: filter the elements that meet a condition (pandas uses the same idea) |

````python
from array import array

int_arr = array('i', [10, 20, 30])
int_arr.append(40)
print(int_arr, int_arr[1], len(int_arr))

try:
    int_arr.append(2.5)          # arrays enforce a single type
except TypeError as e:
    print("TypeError:", e)
````

````python
import numpy as np

prices = np.array([19.99, 5.49, 3.75, 12.00])
print(prices, prices.dtype, prices.shape)

# Vectorized math: no loop required
print("With 8% tax:", prices * 1.08)
print("Total:", prices.sum(), "| Mean:", prices.mean().round(2))

# Compare with a plain list
lst = [19.99, 5.49, 3.75, 12.00]
print(lst * 2)                 # a list REPEATS
print(prices * 2)              # an array MULTIPLIES each element
````

````python
# 2-D arrays (rows x columns), the foundation of tables
matrix = np.array([[1, 2, 3],
                   [4, 5, 6]])
print(matrix)
print("shape:", matrix.shape, "| ndim:", matrix.ndim, "| size:", matrix.size)
print("Element at row 1, col 2:", matrix[1, 2])
print("First column:", matrix[:, 0])
print("Column sums (axis=0):", matrix.sum(axis=0))
print("Row sums (axis=1):   ", matrix.sum(axis=1))

print(np.arange(0, 10, 2))
print(np.linspace(0, 1, 5))
print(np.zeros((2, 3)))
print(np.arange(6).reshape(2, 3))
````

````python
# Boolean masking: filtering data
temps = np.array([12.1, 25.3, 30.8, 18.4, 33.0])
hot = temps > 25
print(hot)               # an array of True/False values
print(temps[hot])        # only the hot days
print(temps[(temps > 15) & (temps < 31)])   # combine conditions with & and | (use parentheses!)
````

**Solution 11: Error Handling**

````python
def get_field(record, key):
    try:
        return int(record[key])
    except KeyError:
        print(f"Missing key: {key}")
        return 0
    except ValueError:
        print(f"Cannot convert {record[key]!r} to int")
        return 0
    finally:
        print("lookup done")

for rec in [{"views": "120"}, {"views": "lots"}, {"likes": "5"}]:
    print("Result:", get_field(rec, "views"))
````

**Capstone Solution**

````python
import json, csv
import numpy as np
import pandas as pd

raw_json = '''
{"products": [
  {"name": "  wireless mouse ", "price": "$24.99", "category": "Electronics"},
  {"name": "COFFEE MUG", "price": "$9.50", "category": "Kitchen"},
  {"name": "standing desk", "price": "$1,249.00", "category": "Furniture"},
  {"name": "usb-c cable", "price": "N/A", "category": "electronics"},
  {"name": "desk lamp", "price": "$39.00", "category": "FURNITURE"},
  {"name": " french press", "price": "$31.25", "category": "kitchen"}
]}
'''

# 1. Parse
data = json.loads(raw_json)

# 2. Cleaning function
def parse_record(rec):
    try:
        price = float(rec.get("price", "").replace("$", "").replace(",", ""))
    except ValueError:
        price = None
    return {
        "name": rec.get("name", "").strip().title(),
        "price": price,
        "category": rec.get("category", "").lower(),
    }

# 3. Clean, skipping invalid prices
clean = []
for rec in data["products"]:
    parsed = parse_record(rec)
    if parsed["price"] is None:
        print("Skipping:", parsed["name"])
        continue
    clean.append(parsed)

# 4. Unique categories
categories = {p["category"] for p in clean}
print("Categories:", categories)

# 5. NumPy statistics
prices = np.array([p["price"] for p in clean])
print(f"Average price: ${prices.mean():.2f} | Max price: ${prices.max():.2f}")

# 6. Count per category
counts = {}
for p in clean:
    counts[p["category"]] = counts.get(p["category"], 0) + 1
print("Counts:", counts)

# 7. Save to CSV and load with pandas
with open("clean_products.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.DictWriter(f, fieldnames=["name", "price", "category"])
    writer.writeheader()
    writer.writerows(clean)

pd.read_csv("clean_products.csv")
````
