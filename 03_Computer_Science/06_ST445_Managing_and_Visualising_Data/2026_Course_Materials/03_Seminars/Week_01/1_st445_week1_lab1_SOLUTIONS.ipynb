🔵 **Exercise 4: where is this change?** `git status --short` prints two characters before each file name. The first shows the **staging area**, the second the **working directory**:

| Code | Meaning |
|---|---|
| `??` | untracked: a new file Git is not tracking yet |
| ` M` (space, M) | modified in the working directory, not staged |
| `M ` (M, space) | staged, ready to commit |
| `MM` | staged, then modified again after `git add` |
| `A ` | a new file, staged |

Write a function `file_stage(code)` that returns exactly one of `"untracked"`, `"not staged"`, `"staged"`, `"staged and modified again"`, `"new file staged"`, or `"unknown"` for any other code. This is Quiz 3 in code form: in the `MM` case, only the version from the time of `git add` goes into the next commit.

````python
def file_stage(code):
    if code == "??":
        return "untracked"
    elif code == " M":
        return "not staged"
    elif code == "M ":
        return "staged"
    elif code == "MM":
        return "staged and modified again"
    elif code == "A ":
        return "new file staged"
    else:
        return "unknown"
````

🔵 **Exercise 5.** Write a function `conflict_versions(text)` that reads a file containing **one** conflict and returns a tuple `(current, incoming)`: the lines between `<<<<<<<` and `=======`, and the lines between `=======` and `>>>>>>>`, each joined with `"\n"`. For `readme` above it should return `("# Housing affordability in England", "# London A&E waiting times")`.

Hint: loop over the lines with a variable that records which part you are in (`"before"`, `"current"`, `"incoming"`, `"after"`).

````python
def conflict_versions(text):
    part = "before"
    current, incoming = [], []
    for line in text.splitlines():
        if line.startswith("<<<<<<<"):
            part = "current"
        elif line.startswith("=======") and part == "current":
            part = "incoming"
        elif line.startswith(">>>>>>>"):
            part = "after"
        elif part == "current":
            current.append(line)
        elif part == "incoming":
            incoming.append(line)
    return "\n".join(current), "\n".join(incoming)
````

## 6. Functions

A function packages code under a name. It takes **parameters**, can have **default values**, and **returns** a result. A **docstring** in triple quotes says what it does; `help(function_name)` shows it.

````python
# 🟢 Worked example: parameters, defaults, keyword arguments, docstrings
def to_float(raw, missing=("", "NA", "N/A", "NULL", "-999", ".")):
    """Convert a raw string to a float, or return None if it is a missing marker."""
    raw = raw.strip()
    if raw in missing:
        return None
    return float(raw)

print(to_float(" 37.9 "))
print(to_float("-999"))
print(to_float("-999", missing=("",)))   # keyword argument overrides the default
help(to_float)
````

````python
# 🟢 Worked example: a function that returns several values, and scope
def summarise(values):
    """Return the count, minimum and mean of the non-missing values."""
    present = [v for v in values if v is not None]
    return len(present), min(present), sum(present) / len(present)

n, low, mean = summarise([37.9, None, 36.6, 38.4])
print(f"n={n}, min={low}, mean={mean:.2f}")
# 'present' existed only inside the function: it is a local variable.
````

🟡 **Complete the code.** Finish `convert_temperature` so that it converts Celsius to Fahrenheit by default, and Fahrenheit to Celsius when `to="C"`.

````python
def convert_temperature(value, to="F"):
    """Convert between Celsius and Fahrenheit, rounded to one decimal place."""
    if to == "F":
        result = value * 9 / 5 + 32
    elif to == "C":
        result = (value - 32) * 5 / 9
    else:
        raise ValueError("to must be 'F' or 'C'")
    return round(result, 1)

check(convert_temperature(37.9) == 100.2, "The default should convert to Fahrenheit.")
check(convert_temperature(100.2, to="C") == 37.9, "Fahrenheit to Celsius is (F - 32) * 5 / 9.")
````

🔵 **Exercise 6: a commit-message checker.** Write `check_commit_message(msg)` that returns a **list of problems** (an empty list means the message is good). Report a problem when:

1. the message is empty or only spaces: `"empty"`
2. the first line is longer than 50 characters: `"too long"`
3. the first word, in lower case, is one of `update`, `fix`, `stuff`, `changes`, `wip` **and** the message has only one word: `"too vague"`
4. the first character is not upper case: `"not capitalised"`
5. the first line ends with a full stop: `"ends with full stop"`

For an empty message, return `["empty"]` and check nothing else.

````python
def check_commit_message(msg):
    """Return a list of problems with a commit message; empty if it is fine."""
    if not msg.strip():
        return ["empty"]
    problems = []
    first_line = msg.strip().splitlines()[0]
    words = first_line.split()
    if len(first_line) > 50:
        problems.append("too long")
    if words[0].lower() in {"update", "fix", "stuff", "changes", "wip"} and len(words) == 1:
        problems.append("too vague")
    if not first_line[0].isupper():
        problems.append("not capitalised")
    if first_line.endswith("."):
        problems.append("ends with full stop")
    return problems
````

## 7. Methods and classes

You have been calling methods on built-in objects (`text.strip()`, `hashes.append(...)`). A **class** lets you define your own kind of object with its own data (**attributes**) and methods. `self` is the object the method is called on.

````python
# 🟢 Worked example: a Commit class
class Commit:
    """One commit: an id, an author and a message."""

    def __init__(self, hash_, author, message):
        self.hash = hash_
        self.author = author
        self.message = message

    def short(self):
        """Return the one-line form used by git log --oneline."""
        return f"{self.hash[:7]} {self.message}"

    def __repr__(self):
        return f"Commit({self.hash[:7]!r}, {self.author!r})"

c1 = Commit("e18d4f7aa", "Lin", "Parse admission dates as ISO 8601 in UTC")
print(c1.short())
print(c1.author)
print(c1)
````

🟡 **Complete the code.** A `Branch` holds a name and a list of commits. Complete `add_commit` and `log`, which returns the one-line forms, newest first.

````python
class Branch:
    def __init__(self, name):
        self.name = name
        self.commits = []

    def add_commit(self, commit):
        self.commits.append(commit)

    def log(self):
        return [c.short() for c in reversed(self.commits)]

main = Branch("main")
main.add_commit(Commit("a41f9c2", "Priya", "Add module repository template"))
main.add_commit(Commit("7d03e11", "Sam", "Add data source and licence to README"))
print("\n".join(main.log()))

check(main.log() == ["7d03e11 Add data source and licence to README", "a41f9c2 Add module repository template"],
      "Append to the list; in log, call each commit's short() method.")
````

🔵 **Exercise 7: model the three places a change can be.** Write a class `MiniRepo` that imitates Git's working directory, staging area and history:

* `__init__(self)` creates three attributes: `working` and `staged` (empty dictionaries mapping file name to content) and `history` (an empty list).
* `edit(self, filename, content)` puts the content into `working`.
* `add(self, filename)` copies the **current** working content of that file into `staged`.
* `commit(self, message)` appends a dictionary `{"message": message, "files": <a copy of staged>}` to `history`, then empties `staged`. If nothing is staged, raise `ValueError("nothing to commit")`.
* `status(self, filename)` returns `"not staged"` if the working content differs from the staged or last-committed version and it is not staged, `"staged"` if the staged content equals the working content, and `"staged and modified again"` if it is staged but the working content has changed since.

The test reproduces Quiz 3, question 2.

````python
class MiniRepo:
    def __init__(self):
        self.working = {}
        self.staged = {}
        self.history = []

    def edit(self, filename, content):
        self.working[filename] = content

    def add(self, filename):
        self.staged[filename] = self.working[filename]

    def commit(self, message):
        if not self.staged:
            raise ValueError("nothing to commit")
        self.history.append({"message": message, "files": dict(self.staged)})
        self.staged = {}

    def status(self, filename):
        if filename in self.staged:
            if self.staged[filename] == self.working[filename]:
                return "staged"
            return "staged and modified again"
        return "not staged"
````

````python
# Test for Exercise 7
repo = MiniRepo()
repo.edit("clean.py", "version 1")
check(repo.status("clean.py") == "not staged", "A new edit that has not been added is not staged.")
repo.add("clean.py")
check(repo.status("clean.py") == "staged", "After add, the file is staged.")
repo.edit("clean.py", "version 2")
check(repo.status("clean.py") == "staged and modified again", "Edited after add.")
repo.commit("Add cleaning script")
check(repo.history[-1]["files"]["clean.py"] == "version 1", "The commit records the version from the time of add (Quiz 3, Q2).")
try:
    repo.commit("Empty commit")
    check(False, "Committing with nothing staged should raise ValueError.")
except ValueError:
    check(True)
````

## 8. Errors and exceptions

When something goes wrong Python **raises** an exception. `try` and `except` let your code respond instead of stopping. Catch the specific exception you expect, not every error.

````python
# 🟢 Worked example: try, except, else, raise
def safe_float(raw):
    """Return a float, or None if raw cannot be converted."""
    try:
        value = float(raw)
    except ValueError:
        return None
    else:
        return value

print([safe_float(x) for x in ["37.9", "high", "", "38"]])

def check_triage(level):
    if level not in range(1, 6):
        raise ValueError(f"triage must be 1 to 5, got {level}")
    return level

try:
    check_triage(7)
except ValueError as err:
    print("Caught:", err)
````

🔵 **Exercise 8.** Write `parse_timestamp(stamp)` that turns an ISO 8601 string such as `"2026-10-05T14:30Z"` into a `datetime` object in UTC, using `datetime.fromisoformat`. Replace a trailing `Z` with `+00:00` first. If the string is not a valid timestamp, or has no time zone, raise `ValueError` with a helpful message. (A date-time with no zone is ambiguous: the lecture's rule is always to keep the zone.)

````python
from datetime import datetime, timezone

def parse_timestamp(stamp):
    """Parse an ISO 8601 timestamp with a time zone into a UTC datetime."""
    if stamp.endswith("Z"):
        stamp = stamp[:-1] + "+00:00"
    try:
        dt = datetime.fromisoformat(stamp)
    except ValueError:
        raise ValueError(f"not an ISO 8601 timestamp: {stamp!r}")
    if dt.tzinfo is None:
        raise ValueError(f"timestamp has no time zone: {stamp!r}")
    return dt.astimezone(timezone.utc)
````

````python
# Test for Exercise 8
dt = parse_timestamp("2026-10-05T14:30Z")
check(dt.hour == 14 and dt.utcoffset().total_seconds() == 0, "Replace Z with +00:00 before parsing.")
check(parse_timestamp("2026-10-05T15:30+01:00").hour == 14, "Convert to UTC with astimezone(timezone.utc).")
for bad in ["05/10/2026", "2026-10-05T14:30"]:
    try:
        parse_timestamp(bad)
        check(False, f"{bad!r} should raise ValueError.")
    except ValueError:
        check(True)
````

🔵 **Exercise 9: write a data dictionary.** Write a function `write_data_dictionary(columns, path)` that takes a dictionary mapping column names to descriptions and writes a Markdown table to `path`:

```
| column | description |
|---|---|
| id | Patient identifier, stored as text |

```

Return the number of columns written. Then call it with `data_columns` below to create `data_dictionary.md`. It goes in the `data/` folder of your personal repository at checkpoint E: a data dictionary is part of the README quality criterion in the repository health check.

````python
data_columns = {
    "id": "Patient identifier, stored as text to keep leading zeros",
    "country_code": "ISO 3166-1 alpha-2 code; NA means Namibia",
    "temp_c": "Body temperature in degrees Celsius; missing if not recorded",
    "admitted_at": "Admission time, ISO 8601 in UTC",
}

def write_data_dictionary(columns, path):
    lines = ["| column | description |", "|---|---|"]
    for name, description in columns.items():
        lines.append(f"| {name} | {description} |")
    with open(path, "w", encoding="utf-8") as f:
        f.write("\n".join(lines) + "\n")
    return len(columns)

write_data_dictionary(data_columns, "data_dictionary.md")
````
