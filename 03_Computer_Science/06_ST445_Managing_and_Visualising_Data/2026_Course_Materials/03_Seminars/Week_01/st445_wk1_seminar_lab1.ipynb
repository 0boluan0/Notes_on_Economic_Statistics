## 6. Functions

A function packages code under a name. It takes **parameters**, can have **default values**, and **returns** a result. A **docstring** in triple quotes says what it does; `help(function_name)` shows it.

````python
# 🟢 Worked example: parameters, defaults, keyword arguments, docstrings
def to_float(raw, missing=("", "NA", "N/A", "NULL", "-999", ".")):
    """Convert a raw string to a float, or return None if it is a missing marker."""
    raw = raw.strip()
    if raw in missing:
        return None
    return float(raw)

print(to_float(" 37.9 "))
print(to_float("-999"))
print(to_float("-999", missing=("",)))   # keyword argument overrides the default
help(to_float)
````

````python
# 🟢 Worked example: a function that returns several values, and scope
def summarise(values):
    """Return the count, minimum and mean of the non-missing values."""
    present = [v for v in values if v is not None]
    return len(present), min(present), sum(present) / len(present)

n, low, mean = summarise([37.9, None, 36.6, 38.4])
print(f"n={n}, min={low}, mean={mean:.2f}")
# 'present' existed only inside the function: it is a local variable.
````

## 7. Methods and classes

You have been calling methods on built-in objects (`text.strip()`, `hashes.append(...)`). A **class** lets you define your own kind of object with its own data (**attributes**) and methods. `self` is the object the method is called on.

````python
# 🟢 Worked example: a Commit class
class Commit:
    """One commit: an id, an author and a message."""

    def __init__(self, hash_, author, message):
        self.hash = hash_
        self.author = author
        self.message = message

    def short(self):
        """Return the one-line form used by git log --oneline."""
        return f"{self.hash[:7]} {self.message}"

    def __repr__(self):
        return f"Commit({self.hash[:7]!r}, {self.author!r})"

c1 = Commit("e18d4f7aa", "Lin", "Parse admission dates as ISO 8601 in UTC")
print(c1.short())
print(c1.author)
print(c1)
````

## 8. Errors and exceptions

When something goes wrong Python **raises** an exception. `try` and `except` let your code respond instead of stopping. Catch the specific exception you expect, not every error.

````python
# 🟢 Worked example: try, except, else, raise
def safe_float(raw):
    """Return a float, or None if raw cannot be converted."""
    try:
        value = float(raw)
    except ValueError:
        return None
    else:
        return value

print([safe_float(x) for x in ["37.9", "high", "", "38"]])

def check_triage(level):
    if level not in range(1, 6):
        raise ValueError(f"triage must be 1 to 5, got {level}")
    return level

try:
    check_triage(7)
except ValueError as err:
    print("Caught:", err)
````
